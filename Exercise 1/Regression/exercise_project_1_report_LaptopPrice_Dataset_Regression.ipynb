  <li>Weight is non-linear. Its correlation is weak, but Phik, mutual information, RFE and the decision tree rank it high, so it should be binned or used with tree-based models.</li>
  <li>Os and Company_Apple are redundant (Apple laptops run Mac), which showed as extreme VIF values. Mac and Others should be merged into one "is Windows" column. Cpu_brand and TypeName dummies overlap moderately.</li>
  <li>Os and TouchScreen are heavily imbalanced, and Company has 19 categories with many rare ones, so rare brands should be grouped into "Other".</li>
  <li>TouchScreen, HDD and the rare Company categories are weak and can be dropped.</li>
  <li>Noise (MAD on Weight) and outliers (Isolation Forest) are mild. They are real high-end or unusual laptops and not obvious errors, so they can be kept, especially for tree-based models.</li>
  <li>The price groups overlap heavily on single features (HDD, Weight, Ppi), but together Ram, SSD and Cpu_brand separate them well enough.</li>
</ul>
