### y-data with extra features

```python
from ydata_profiling import ProfileReport
from ydata_profiling.config import Settings
import pandas as pd
import numpy as np


df = pd.read_csv("../heart.csv")

# Create configuration
config = Settings()

# Set the skewness threshold to whichever you prefer.
config.vars.num.skewness_threshold = 2

# Enable correlations
config.correlations["pearson"].calculate = True
config.correlations["phi_k"].calculate = True

# Generate the profile report
report = ProfileReport(
    df,
    title='Heart Failure Prediction Dataset / Advanced Data Analytics, 2026',
    config=config
)

# Save report
report.to_file("ydata_heart_failure_prediction_dataset.html")
```
